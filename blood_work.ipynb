```python
from dotenv import load_dotenv



from langchain_groq import ChatGroq


# from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

llm = ChatGroq(model = "openai/gpt-oss-20b", reasoning_format = "parsed")

# llm = ChatGoogleGenerativeAI(model = "gemini-2.5-flash")
 
```

```python
with open("blood_work_db.txt", "r") as f:
    blood_report = f.read()
print(blood_report)
```

```python
prompt =  f"""

You are medical ai assitant and you are going to extract all the values from the blood report i have attached below and classify each one as normal , mid and high based on the reference provided in the file.

Structure you response as :

- Test name | Status : High/mid/normal | Reference : range | actual value 

Blood report :
{blood_report}

"""

response = llm.invoke(prompt)
values = response.text
print ( "=== stage 1 : Extracted values ===")
print(values)
```

```python
diet_prompt = f"""

You are ai clinical nutritionist specialized in giving amazing indian diet food suggestion.

Take the blood word report as data and suggest a deit plan which includes a structure of :

Structure : Food options in tabular column (neat minimal format)  | What not to eat| Food habits to do and not to do {blood_report} use emojies and neat format to display

At last tell me How much cost it took for processing this prompt tokens,k values and embeddings and whihc llm is called and time took and tokens used and generated and remaining balance and which api in a tabular column order

"""

response = llm.invoke(diet_prompt)
print(response.text)
```
